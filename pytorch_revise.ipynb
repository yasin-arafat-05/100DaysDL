# `1. View:`

<br>

```python
# 1. view:
import numpy as np
import torch

X = torch.rand(3,4)

print(X)
# we have total 12 element, the view do that we can build any shpae of matrix
# that excetly contain 12 element, like, (6,2)=12, (1*12),
print(X.view(1,12))
print(X.view(12,1))
print(X.view(2,6))
print(X.view(6,2))
```

# `2. Transpose, Contigious:`

<br>

```python

# transpose:
X = torch.rand((4,4))
X = X.view(2,8)
print(X)
# dim=0 and dim=1 or dim=-1,dim=-2
print(X.transpose(0,1))
```

```python
X = torch.rand((3,4,4))
print(X)
print()
# dim0=3, dim1=4,dim2=4
print(X.transpose(1,2))
```

```python


# after doing: transpose our colacelent memory accss is gone:
# colacelent memory access
x = torch.rand(4,4)
print(x.is_contiguous())
x = x.transpose(0,1)
print(x.is_contiguous())
# our contigious funtion fix this
x = x.contiguous()
print(x.is_contiguous())
```

# `4. Interpolate:`

<br>

```python


import torch 
import torch.nn.functional as F

demo_img1d = torch.tensor(
    data=[
        [1,2,2,2,1],
        [1,2,2,2,1],
        [1,2,2,2,1],
        [1,2,2,2,1],
        [1,2,2,2,1],
    ],
    dtype=torch.float32
)
demo_img1d = demo_img1d.unsqueeze(0).unsqueeze(0)

print(demo_img1d.shape)
print(demo_img1d)
print()
print(F.interpolate(demo_img1d,size=(10,10),mode="nearest"))
```
